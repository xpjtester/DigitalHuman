## Extract a 10-second motion template
Change `START` to choose a good speaking section with a fairly visible face.

```
import subprocess, os
START=0
DURATION=10
template='/content/digitalhuman/private/template_10s.mp4'
subprocess.run(['ffmpeg','-y','-ss',str(START),'-i',source_path,'-t',str(DURATION),'-c:v','libx264','-crf','18','-c:a','aac',template],check=True)
print(template)
```

```
from IPython.display import Video
Video(template, embed=True, width=360)
```
